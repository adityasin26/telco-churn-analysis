```python

df.groupby("Contract")["Churn"].value_counts(normalize=True).unstack() *100
```

```python
df.groupby("Churn")["tenure"].mean()
```

```python
df["tenure_group"] = pd.cut(
    df["tenure"],
    bins=[0, 12, 24, 48, 72],
    labels=["0-12 months", "13-24 months", "25-48 months", "49-72 months"],
    include_lowest=True
)

df.groupby("tenure_group")["Churn"].value_counts(normalize=True).unstack()*100
```

```python
pd.crosstab(df["tenure_group"], df["Contract"], values=df["Churn"].eq("Yes"), aggfunc="mean").round(3) * 100
```
